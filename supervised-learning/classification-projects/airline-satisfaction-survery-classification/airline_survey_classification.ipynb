```python
import pandas as pd
import numpy as np
import seaborn as sns
from matplotlib import pyplot as plt 
from sklearn.model_selection import train_test_split
```

#### Data Import and Inspection

```python
survey_df = pd.read_csv('data/passengers_survey_responses.csv')
print(f"The Size of the dataset: {survey_df.shape}")
print(f"Data Sample: ")
survey_df.head()
```

```python
survey_df.describe()
```

```python
survey_df.info()
```

```python
survey_df.columns = survey_df.columns.str.replace(' ','_').str.lower()
```

```python
survey_df.columns
```

```python
# Map the Output class from Bool -> Int

survey_df['satisfaction'] = survey_df['satisfaction'].map({False: 0, True: 1})
```

```python
survey_df.dtypes
```

- Performing the Train_Test split so that the model has less inference of our test data while training

```python
X = survey_df.drop('satisfaction', axis=1)
y = survey_df['satisfaction']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```
